```python
driver = "org.postgresql.Driver"
url    = "jdbc:postgresql://<rds-endpoint>:5432/your_db?sslmode=require"

props = {
  "user": dbutils.secrets.get("scope", "pg_user"),
  "password": dbutils.secrets.get("scope", "pg_password"),
  "driver": driver
}

# Read a table into Spark
df = (spark.read
      .format("jdbc")
      .option("url", url)
      .option("dbtable", "public.your_table")
      .options(**props)
      .load())

df.display()

# Write back to a new table
(df.write
  .format("jdbc")
  .option("url", url)
  .option("dbtable", "public.your_table_out")
  .options(**props)
  .mode("overwrite")
  .save())
```
